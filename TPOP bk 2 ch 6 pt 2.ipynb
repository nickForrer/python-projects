```python
#using class methods


#def methodname(self(param1, param2))

import datetime as dt

class Member:
   
    #create class variable - here default num of free days
    free_days = 365

    '''create a new member from uname and fname'''
    

    def __init__(self,username,fullname):
        
        #define attributes and give them values
        self.username = username
        self.fullname = fullname
        
        #set default date joined, to todays date
        self.date_joined = dt.date.today()
        
        #set an expiration date
        self.free_expires = dt.date.today() + dt.timedelta(days = Member.free_days)

        #set is active to True initially
        self.is_active = True
        
    #method to activate True or deactivate False account
    def activate(self, yesno):
        '''True for Active, False to make inactive'''
        self.is_active = yesno
        
    #class methods follow @classmethod decorator and refer to cls rather than self
    @classmethod
    def setfreedays(cls,days):
        cls.free_days = days

    
    
    #create an instance of the member class named new guy
new_guy = Member('Rambo', 'Rocco Moe')

print(new_guy.date_joined)
print(new_guy.free_expires)
new_guy.activate(False)
print(new_guy.is_active)
Member.setfreedays(90)
print(new_guy.free_expires)
```

```python
#static methods - not tied to class or object

import datetime as dt

class Member:
   
    #create class variable - here default num of free days
    free_days = 365

    '''create a new member from uname and fname'''
    

    def __init__(self,username,fullname):
        
        #define attributes and give them values
        self.username = username
        self.fullname = fullname
        
        #set default date joined, to todays date
        self.date_joined = dt.date.today()
        
        #set an expiration date
        self.free_expires = dt.date.today() + dt.timedelta(days = Member.free_days)

        #set is active to True initially
        self.is_active = True
        
    #method to activate True or deactivate False account
    def activate(self, yesno):
        '''True for Active, False to make inactive'''
        self.is_active = yesno
        
    #class methods follow @classmethod decorator and refer to cls rather than self
    @classmethod
    def setfreedays(cls,days):
        cls.free_days = days

    @staticmethod
    def current_time():
        now = dt.datetime.now()
        return f'{now:%I:%M %p}'
    
    
    #create an instance of the member class named new guy
new_guy = Member('Rambo', 'Rocco Moe')

print(Member.current_time())

```

```python
#inherentence parent/base class and subclasses inheret all attributes and behaviors

#class is used for all kinds of people

import datetime as dt

#base class is used for all kinds of members

class Member:
   
    '''the Member class attributes and methods are for everyone'''
    #by default a new account expires after one year - class attribute
    expiry_days = 365

    
    #initialise a member object

    def __init__(self,firstname,lastname):
        
        #define attributes and give them values
        self.firstname = firstname
        self.lastname = lastname
        
        #default secret code is nothing
        self.secret_code = ''
        
        #set default date joined, to todays date
        self.date_joined = dt.date.today()
        
        #set an expiration date
        self.expiry_date = dt.date.today() + dt.timedelta(days = self.expiry_days)

    #method in the base class
    def showexpiry(self):
        return f'{self.firstname} {self.lastname} expires on {self.expiry_date}'
    
#create an instance of the member class
Joe = Member('Joe', 'Anybody')
print(Joe.firstname)
print(Joe.lastname)
print(Joe.expiry_date)

#print(Member.current_time())

#subclass for admins

class Admin(Member):
    #Admin accounts dont expire for 100 years
    expiry_days = 365*100

    def __init__(self, firstname, lastname, secret_code):
        super().__init__(firstname, lastname)
        self.secret_code = secret_code
        
    def get_status(self):
        return f'{self.firstname} is an Admin'
    
#subclass for users

class User(Member):
    def get_status(self):
        return f'{self.firstname} is a regular user'

Ann = Admin('Annie', 'Angst', 'Presto')

print(Ann.firstname)
print(Ann.lastname)
print(Ann.expiry_date)
print(Ann.secret_code)

Uli = User('Uli', 'Ungula')

print(Uli.firstname)
print(Uli.lastname)
print(Uli.expiry_date)
print(Uli.secret_code)

print(Ann.showexpiry())
print(Uli.showexpiry())

print(Ann.get_status())

help(Admin)


```
