```python
#TABLE PRINTER

tableDataExample = [['apples', 'oranges', 'cherries', 'banana'],
             ['Alice', 'Bob', 'Carol', 'David'],
             ['dogs', 'cats', 'moose', 'goose']]

def printTable(tableData):
    #code to find and store maxlength string
    colWidths = []
    
    for subList in tableData:
        lengthStore = []
        for entry in subList:
            lengthStore.append(len(entry))
    
        largestNum = max(lengthStore)
        colWidths.append(largestNum)  
    #code to print strings in a table, rjust by maxlength in each column
    for i in range(len(tableData) +1):
        for j in range(len(tableData)):
            print(tableData[j][i].rjust(colWidths[j]), end = ' ')
        print('')

printTable(tableDataExample)

```
