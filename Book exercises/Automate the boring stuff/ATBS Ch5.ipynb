```python
#CHESS DICTONARY VALIDATOR

#is valid board

givenBoard1 = {'1a': 'wking', '8h': 'bking' , '2e': 'bking'}

def isValidChessBoard(givenBoard):

    pcList = ['pawn', 'bishop', 'king', 'rook', 'knight', 'queen']

    colour = ['w', 'b']

    row = ['a','b','c','d','e','f','g','h']

    wkingNum = 0
    wpcNum = 0
    wpwnNum = 0
    bkingNum = 0
    bpcNum = 0
    bpwnNum = 0
    
    for key, value in givenBoard.items():

        if int(key[0]) not in range(1,9):
            return False
        if key[1].lower() not in row:
            return False
        if value[1:].lower() not in pcList:
            return False
        if value[0].lower() not in colour:
            return False
        
        if value.lower() == 'wking':
            wkingNum += 1
        if value.lower() == 'bking':
            bkingNum += 1
        if value[0].lower() == 'w':
            wpcNum += 1
        if value[0].lower() == 'b':
            bpcNum += 1
        if value.lower() == 'wpawn':
            wpwnNum += 1
        if value.lower() == 'bpawn':
            bpwnNum += 1
    
    if wpcNum > 16 or bpcNum > 16 or bkingNum > 2 or wkingNum > 2 or wpwnNum > 8 or bpwnNum > 8:
        return False
        
    else:
        return True
    
isValidChessBoard(givenBoard1)
            
```

```python
#Add To Fantasy Game Inventory

dragonLoot = ['gold coin', 'dagger', 'gold coin', 'gold coin', 'ruby']

loot = {'arrow': 12, 'gold coin': 42, 'rope': 1, 'torch': 6, 'dagger': 1}

def displayInventory(inventory):
    print('Inventory: ')
    itemTotal = 0
    for k, v in inventory.items():
        print ( str(v) + ' ' + k)
        itemTotal += v
    print( f'Total number of items: {itemTotal}') 

def addToInventory(inventory, addedItems):    
    for item in addedItems:
        if item in inventory.keys():
            inventory[item] += 1
        if item not in inventory.keys():
            inventory[item] = 1
    
addToInventory(loot, dragonLoot)    
displayInventory(loot) 
```
