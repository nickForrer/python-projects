```python
#COMMA CODE

spam = ['apples', 'bananas', 'tofu', 'cats']

spam2 = ['bob', 'garry']

def CommaCode(givenList):
    if len(givenList) == 0:
        print('the list is empty')
    elif len(givenList) == 1:
        print(givenList[0])
    elif len(givenList) == 2:
        string = ' and '.join(givenList)
        print(string)
    else:
        string = ', '.join(givenList[0:len(givenList) - 1]) + ' and ' + givenList[-1]
        print(string)
    
CommaCode(spam)
```
