```python
#COLLATZ SEQUENCE

def collatz(number):
    if number % 2 == 0:
        result = number // 2
        print (result)
        return result
    else:
        result = 3 * number + 1
        print(result)
        return result

try:
    
    userInput = int(input('Enter number: '))
    
    res = collatz(userInput)

    while True:
        
        res = collatz(res)
        if res == 1:
            break
            
except ValueError:

    print('You must enter a number!')
    
```
