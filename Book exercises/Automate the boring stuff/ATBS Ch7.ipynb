```python
#datedetection

text = 'bob is your uncle and the day is 31/04/2003 another day is 04/12/1899'

import re

dateRegex = re.compile (r'''[0-3][0-9]/[0-1][0-9]/\d{4}''')

results = dateRegex.findall(text) #findall method returns a list of results

#print regex results
print('The results found are: ')
for i in results:
    print(i)

#function that checks if a year is leap
def isLeap(year):
    if year % 4 == 0 and year % 100 != 0:
            return True
    elif year % 4 == 0 and year % 100 == 0 and year % 400 == 0:
            return True
    else:
        return False

#iterate through list and assign variable to date month and year
for i in range(len(results)):
    date = results[i][0:2]
    month = results[i][3:5]
    year = results[i][6:11]
    
    shortMonths = [4,6,9,11]
    
#check if valid date, month and year
    
    if int(date) > 31:
        print(f'{results[i]} is an invalid date. There are no more than 31 days in a month')
    elif int(month) > 12:
        print(f'{results[i]} is an invalid date. There are only 12 months in a year.')
    elif int(month) in shortMonths and int(date) > 30:
        print(f'{results[i]} is an invalid date. There are only 30 days in that month.')
    elif int(month) == 2:
        if isLeap(int(year)) and int(date) > 29:
            print(f'{results[i]} is an invalid date. Feb only has 29 days in Leap years')
        elif not isLeap(int(year)) and int(date) > 28:
            print(f'{results[i]} is an invalid date. Feb only has 28 days in non Leap years')
```

```python
#Strong password detection



import re

usrpwd = input('please enter a password: ')


def strongPassword(password):
    
    #use regex to asess password could use groups for this?
    
    lcRegex = re.compile(r'[a-z]') #regex goes here
    lcResults = lcRegex.findall(password)
    upcRegex = re.compile(r'[A-Z]')
    upcResults = upcRegex.findall(password)
    digRegex = re.compile(r'[0-9]')
    digResults = digRegex.findall(password)
    #print(upcResults, lcResults, digResults)
    # condition at least eight characters long.
    if len(password) < 8:
            print('Error: Password too short')
    #contains both uppercase and lowercase characters
    if len(upcResults) == 0:
            print('Error: Password must contain at least one uppercase letter')
    if len(lcResults) == 0:
            print('Error: Password must contain at least one lowercase letter')
    if len(digResults) == 0:
            print('Error: Password must contain at least one digit')
    else:
        print('No errors detected. Password is strong')

strongPassword(usrpwd)
```
