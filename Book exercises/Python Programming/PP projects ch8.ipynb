```python
#Vingenere cipher
dic = ' abcdefghijklmnopqrstuvwxyz'
phrase = "cat"
var = "Feed Me"
newvar = var.lower()

out = []

for ch in phrase:
    num1 = dic.find(ch)
    print(num1)
    out.append(num1)
    #num = dic[oldpos]
print(out)

code = ""

for i, ch in enumerate(newvar):
    oldpos = dic.find(ch)
    keyshift = out[i % len(out)]
    newpos = (oldpos + keyshift) % len(dic)
    newch = dic[newpos]
    code  += newch

print(code)

decode = ""
for i, ch in enumerate(code):
    oldpos = dic.find(ch)
    keyshift = out[i % len(out)]
    newpos = (oldpos - keyshift) % len(dic)
    newch = dic[newpos]
    decode  += newch

print(decode)
```

```python
#national weather

def wChill(T,V):
    return 35.74 + 0.6125*T - 35.75*(V**0.16) + 0.4275*T*(V**0.16)


def main():
    speed = [i*5 for i in range(1,11)]
    #print(speed)
    temp = [i*10 for i in range(-2, 7)]
    #print(temp)

    print(f"{'Temp (°F)':^10} {'Wind (mph)':^12} {'Wind Chill (°F)':^15}")
    print("-" * 37)


    for t in temp: 
        for v in speed:
            wc = wChill(t,v)
            print(f"{t:^10} {v:^12} {wc:^15.2f}")

#main()
```

```python
# Pig Latin translator
def firstVowel(word):
    """
    Returns the index of the first vowel (a, e, i, o, u) and the prefix before it.
    """
    vowels = 'aeiou'

    for i, char in enumerate(word):
        if char in vowels:
            return i, word[:i]  # Index of vowel, prefix before it


def transWordPl(word):
    index, rest = firstVowel(word) #Tuple unpacking
    if index == 0:  # Starts with vowel
        return word + "way"
    else:  # Consonant start
        return word[index:] + rest + "ay"  # After vowel + prefix + "ay"

def main():
    prompt = input("Give sentence: ")
    
    atom = prompt.split(" ")    
    final = ""
    for word in atom:
        final += transWordPl(word) + " "

    print(final)

main()
```
