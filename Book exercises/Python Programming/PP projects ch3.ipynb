```python
#vol and surface of a sphere

import math

def main():
    
    rad = float(input("input the radius"))
    vol = rad**3 * math.pi * (3/4)
    area = 4 * math.pi * (rad**2)
    print("the volume is", vol, "and the surface area is", area)
    
main()
```

```python
#pizza cost

def main():
    
    price = float(input("the price is: "))
    diam = float(input("the diameter is: "))
    rad = diam / 2
    area = math.pi * (rad**2)
    output = price/area #make sure this is right way round!
    print("the cost per square inch is: ", output)
    
main()



```

```python
#slope

def main():
    x1,x2,y1,y2 = map(float, input("Enter x1,x2,y1,y2 with space between each: ").split())
    slope = (y2-y1)/(x2-x1)
    dist = math.sqrt((x2-x1)**2 + (y2-y1)**2)
    print("the slope of the line is: ", slope, "the length of the line section is: ", dist)

main()
```

```python
#triangle

def main():
    a, b, c = map(float, input("Enter sides a, b, c with space between each: ").split())
    s = (a + b + c)/2
    area = math.sqrt(s*(s-a)*(s-b)*(s-c))
    print(f"the area is: {area:.2f}")
    
main()
    
```

```python
#ladder

def main():
    height = float(input("enter the height in metres: "))
    angle = float(input("enter the angle in deg"))
    rad = (math.pi/180)*angle
    length = height/(math.sin(rad))
    print(f"the length of the ladder required is {length:.2f} metres ")
    
main()
```

```python
def main():
    n = int(input("give int value: "))
    total = 0
    totalcube = 0
    for i in range(1,n+1):
        cube = i**3
        total = total + i
        totalcube = totalcube + cube
    print(f"the sum of integers up to {n} is: {total} and the sum of cubes is: {totalcube}")
    
main()
    
```

```python
#fib sequence
def main():
    
    f1 = 1
    f2 = 1
    total = 0
    n = int(input("give nth term "))
    for i in range(n-2):
        total = f1 + f2
        f1, f2 = f2, total
        print(total)
    print(total)

main()
```
