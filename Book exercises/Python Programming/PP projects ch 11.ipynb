```python
#6 volleyball regulation

from random import random

def main():
    printIntro()
    probA, probB, n = getInputs()
    winsA, winsB, shutoutsA, shutoutsB = simNSets(n, probA, probB)
    printSummary(winsA, winsB, shutoutsA, shutoutsB)

def printIntro():
    print("This program simulates a volleyball set between two teams")
    print('called "A" and "B" using rally scoring.')
    print("Every rally scores a point, regardless of who serves.")
    print("The abilities of each team are indicated by a probability")
    print("(a number between 0 and 1) that the team wins a rally when serving.")
    print("The first serve alternates between teams each set.")
    print("A team must score at least 25 points and win by 2 points.\n")

def getInputs():
    # Returns the three simulation parameters
    a = float(input("What is the prob. team A wins a rally when serving? "))
    b = float(input("What is the prob. team B wins a rally when serving? "))
    n = int(input("How many sets to simulate? "))
    return a, b, n

def simNSets(n, probA, probB):
    # Simulates n volleyball sets with rally scoring
    # Returns number of wins and shutouts for A and B
    winsA = winsB = shutoutsA = shutoutsB = 0
    for i in range(n):
        # Alternate starting server: A starts even-numbered sets, B starts odd-numbered
        first_server = "A" if i % 2 == 0 else "B"
        scoreA, scoreB = simOneSet(probA, probB, first_server)
        if scoreA > scoreB:
            winsA += 1
            if scoreB == 0:  # Check for shutout (A wins 25-0)
                shutoutsA += 1
        else:
            winsB += 1
            if scoreA == 0:  # Check for shutout (B wins 25-0)
                shutoutsB += 1
    return winsA, winsB, shutoutsA, shutoutsB

def simOneSet(probA, probB, first_server):
    # Simulates one volleyball set with rally scoring
    # first_server indicates who serves first ("A" or "B")
    # Returns final scores for A and B
    serving = first_server
    scoreA = 0
    scoreB = 0
    while not setOver(scoreA, scoreB):
        # Simulate a rally
        if serving == "A":
            if random() < probA:  # A wins the rally
                scoreA += 1  # A scores a point, keeps serving
            else:  # B wins the rally
                scoreB += 1  # B scores a point, takes serve
                serving = "B"
        else:  # serving == "B"
            if random() < probB:  # B wins the rally
                scoreB += 1  # B scores a point, keeps serving
            else:  # A wins the rally
                scoreA += 1  # A scores a point, takes serve
                serving = "A"
    return scoreA, scoreB

def setOver(a, b):
    # a and b represent scores for a volleyball set
    # Returns True if the set is over (at least 25 points and a 2-point lead), False otherwise
    if a >= 25 and a - b >= 2:
        return True
    elif b >= 25 and b - a >= 2:
        return True
    return False

def printSummary(winsA, winsB, shutoutsA, shutoutsB):
    # Prints a summary of wins and shutouts for each team
    n = winsA + winsB
    print("\nSets simulated:", n)
    print(f"Wins for A: {winsA} ({winsA/n:0.1%})")
    print(f"Shutouts by A: {shutoutsA} ({shutoutsA/n:0.1%})")
    print(f"Wins for B: {winsB} ({winsB/n:0.1%})")
    print(f"Shutouts by B: {shutoutsB} ({shutoutsB/n:0.1%})")

if __name__ == '__main__':
    main()
```

```python
#9 craps

from random import randint

def roll_two_dice():
    die1 = randint(1, 6)
    die2 = randint(1, 6)
    return die1 + die2

def game():
    win = 0
    loss = 0
    score = roll_two_dice()
    print(f"Come-out roll: {score}")
    
    # Come-out roll
    if score in [7, 11]:
        win = 1
        print("Win on come-out!")
    elif score in [2, 3, 12]:  # Fixed: 11 -> 12
        loss = 1
        print("Loss on come-out (craps)!")
    else:
        # Point phase
        point = score
        print(f"Point established: {point}")
        while True:
            roll = roll_two_dice()
            print(f"Roll: {roll}")
            if roll == point:
                win = 1
                print("Hit the point! Win!")
                break
            elif roll == 7:
                loss = 1
                print("Rolled a 7. Loss!")
                break
    
    return win, loss

# Simulate 500 games and calculate win probability
total_wins = 0
total_losses = 0
num_games = 500

for i in range(num_games):
    win, loss = game()
    total_wins += win
    total_losses += loss

# Calculate win probability
win_probability = total_wins / num_games

# Output results
print(f"Total Games: {num_games}")
print(f"Total Wins: {total_wins}")
print(f"Total Losses: {total_losses}")
print(f"Win Probability: {win_probability:.4f} ({win_probability*100:.2f}%)")
```

```python
#10 Blackjack

from random import randrange

def choice(myList):
    var1 = randrange(0, len(myList))
    result = myList[var1]
    return result

def draw():
    cards = ['a', 2, 3, 4, 5, 6, 7, 8, 9, 10, 10, 10, 10]
    return choice(cards)

# Simulation parameters
num_hands = 10000  # Number of hands to simulate
bust = 0
stand = 0

for _ in range(num_hands):
    total = 0
    while True:
        card = draw()
        if isinstance(card, int):  # Cards 2–10
            total += card
        elif card == 'a':  # Ace
            if total + 11 <= 21:
                total += 11
            else:
                total += 1
        if total > 21:
            bust += 1
            break
        if total >= 17:
            stand += 1
            break

print(f"Number of busts: {bust}")
print(f"Number of stands: {stand}")
print(f"Bust probability: {bust / num_hands:.4f}")
```

```python
#11 upcard sim

from random import randrange

def choice(myList):
    var1 = randrange(0, len(myList))
    result = myList[var1]
    return result

def draw():
    cards = ['a', 2, 3, 4, 5, 6, 7, 8, 9, 10, 10, 10, 10]
    return choice(cards)

# Simulation parameters
num_hands = 10000  # Number of hands per upcard
upcards = [2, 3, 4, 5, 6, 7, 8, 9, 10, 'a']  # Possible upcards
results = {}

for upcard in upcards:
    bust = 0
    stand = 0
    for _ in range(num_hands):
        # Initialize hand with the upcard
        total = 0
        if isinstance(upcard, int):
            total = upcard
        elif upcard == 'a':
            total = 11  # Start with Ace as 11
        # Draw additional cards until stand or bust
        while total < 17:
            card = draw()
            if isinstance(card, int):
                total += card
            elif card == 'a':
                if total + 11 <= 21:
                    total += 11
                else:
                    total += 1
        if total > 21:
            bust += 1
        else:
            stand += 1
    # Store results for this upcard
    bust_prob = bust / num_hands
    results[upcard] = {'bust': bust, 'stand': stand, 'bust_probability': bust_prob}

# Print results
print("Dealer Bust Probabilities by Upcard:")
for upcard in upcards:
    print(f"Upcard {upcard}: Bust probability = {results[upcard]['bust_probability']:.4f} "
          f"({results[upcard]['bust']} busts, {results[upcard]['stand']} stands)")
```

```python
#13 five of a kind

from random import randrange

def choice(myList):
    var1 = randrange(0, len(myList))
    result = myList[var1]
    return result

def roll():
    die = [1, 2, 3, 4, 5, 6]
    return choice(die)

# Simulation parameters
num_trials = 10000  # Number of trials
five_of_a_kind = 0  # Counter for five-of-a-kind outcomes

for _ in range(num_trials):
    # Roll five dice
    rolls = [roll() for _ in range(5)]
    # Check if all five rolls are the same
    if len(set(rolls)) == 1:  # All rolls are identical if the set has length 1
        five_of_a_kind += 1

# Calculate probability
probability = five_of_a_kind / num_trials

# Print results
print(f"Number of five-of-a-kind: {five_of_a_kind}")
print(f"Total trials: {num_trials}")
print(f"Estimated probability: {probability:.6f}")
print(f"Theoretical probability: {6 / 7776:.6f}")
```

```python
#14 forward and back coin flip

from random import randrange

def choice(myList):
    var1 = randrange(0, len(myList))
    result = myList[var1]
    return result

def coin():
    coin = [1, -1]  # +1 for heads (forward), -1 for tails (backward)
    return choice(coin)

# Simulation parameters
num_trials = 1000
num_flips = 10000
record = []  # Store final position of each trial

for _ in range(num_trials):
    distance = 0  # Start at position 0 for each trial
    for _ in range(num_flips):
        step = coin()  # Flip coin: +1 or -1
        distance += step  # Update position
    record.append(distance)  # Record final position after num_flips

# Calculate averages
avg_position = sum(record) / len(record) if record else 0
avg_abs_distance = sum(abs(x) for x in record) / len(record) if record else 0

# Print results
print(f"Final positions for {num_trials} trials: {record}")
print(f"Average final position: {avg_position:.2f}")
print(f"Average absolute distance: {avg_abs_distance:.2f}")
```

```python
#15 forward and back - most visited square

from random import randrange

def choice(myList):
    var1 = randrange(0, len(myList))
    result = myList[var1]
    return result

def coin():
    coin = [1, -1]  # +1 for heads (forward), -1 for tails (backward)
    return choice(coin)

# Simulation parameters
num_trials = 10
num_flips = 10000
visit_counts = {}  # Dictionary to track visits to each position

for _ in range(num_trials):
    position = 0  # Start at position 0 for each trial
    for _ in range(num_flips):
        # Record visit to current position
        visit_counts[position] = visit_counts.get(position, 0) + 1
        # Take a step
        step = coin()
        position += step

# Print visit counts for each position (sorted by position)
print("Visit counts for each square (position):")
for pos in sorted(visit_counts.keys()):
    print(f"Square {pos}: {visit_counts[pos]} visits")

# Optional: Summary statistics
most_visited = max(visit_counts.items(), key=lambda x: x[1], default=(0, 0))
total_visits = sum(visit_counts.values())
unique_squares = len(visit_counts)
print(f"\nSummary:")
print(f"Most visited square: {most_visited[0]} with {most_visited[1]} visits")
print(f"Total visits: {total_visits}")
print(f"Unique squares visited: {unique_squares}")
print(f"Average visits per square: {total_visits / unique_squares:.2f}")
```

```python
#16 random walk

import math
from random import choice

def pyth(x, y):
    return math.sqrt(x**2 + y**2)

def random_walk_2d(num_steps):
    # Possible steps: right (+1, 0), left (-1, 0), up (0, +1), down (0, -1)
    directions = [(1, 0), (-1, 0), (0, 1), (0, -1)]
    x, y = 0, 0  # Start at origin (0, 0)
    
    for _ in range(num_steps):
        dx, dy = choice(directions)  # Randomly choose a direction
        x += dx  # Update x-coordinate
        y += dy  # Update y-coordinate
    
    return x, y

# Simulation parameters
num_trials = 100
num_steps = 1000
final_positions = []  # Store (x, y) for each trial
distances = []  # Store Euclidean distances from origin

for _ in range(num_trials):
    final_x, final_y = random_walk_2d(num_steps)
    final_positions.append((final_x, final_y))
    distances.append(pyth(final_x, final_y))

# Calculate averages
avg_x = sum(x for x, _ in final_positions) / num_trials
avg_y = sum(y for _, y in final_positions) / num_trials
avg_distance = sum(distances) / num_trials

# Print results
print(f"Number of trials: {num_trials}")
print(f"Number of steps per trial: {num_steps}")
print(f"Average final x-position: {avg_x:.2f}")
print(f"Average final y-position: {avg_y:.2f}")
print(f"Average Euclidean distance from origin: {avg_distance:.2f}")
print(f"Expected average distance (sqrt(N)): {math.sqrt(num_steps):.2f}")
```

```python
#17 random walk any direction

from graphics import *
import random
import time

def random_walk_2d(num_steps, win):
    x, y = 0, 0  # Start at origin (0, 0)
    
    # Draw starting point
    start = Point(x, y)
    start.setFill("red")
    start.draw(win)
    
    for _ in range(num_steps):
        # Generate random angle for step direction
        angle = random.random() * 2 * math.pi
        dx = math.cos(angle)  # x-displacement
        dy = math.sin(angle)  # y-displacement
        # Calculate new position
        new_x, new_y = x + dx, y + dy
        # Draw line from current to new position
        line = Line(Point(x, y), Point(new_x, new_y))
        line.setFill("blue")
        line.draw(win)
        # Update current position
        x, y = new_x, new_y
        # Small delay for animation effect
        time.sleep(0.05)
    
    
    # Draw final point
    end = Point(x, y)
    end.setFill("green")
    end.draw(win)
    
    return x, y

def main():
    # Create graphics window
    win = GraphWin("2D Random Walk", 600, 600)
    win.setCoords(-100, -100, 100, 100)  # Coordinate system: x, y from -100 to 100
    win.setBackground("white")
    
    # Simulate random walk with 1000 steps
    num_steps = 1000
    final_x, final_y = random_walk_2d(num_steps, win)
    
    # Display final position
    print(f"Final position: ({final_x}, {final_y})")
    
    # Keep window open until mouse click
    win.getMouse()
    win.close()

if __name__ == "__main__":
    main()
```
