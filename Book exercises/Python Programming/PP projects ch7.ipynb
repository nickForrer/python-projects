```python
#double invest

def interest(initial, apr, years):
    amount = initial*(1 +apr)**years

    return amount

initial = float(input("enter intial"))
apr = float(input("enter apr"))
total = initial
years = 0
while total < 2*initial:

    years += 1
    total = interest(initial, apr, years)
    print(years, total)
    

print(f"your investment will double in {years} years")

```

```python
#linear regression

from graphics import *
import math

def main():
    
    #draw indpw
    win = GraphWin("linregression", 700,700)
    win.setCoords(-10, -10, 10, 10)
    quit = Rectangle(Point(-8, -8),Point(-6, -7)) 
    quit.setFill("grey")
    quit.draw(win)
    center = quit.getCenter()
    label = Text(center, "DONE")
    label.draw(win)
    
    #setup sentinals
    n = 0
    xtotal = 0
    ytotal = 0
    xytotal = 0
    x2total = 0
    
    #event loop
    while True:
        click = win.getMouse()
        #get click x,y coordinates
        x, y = click.getX(), click.getY()
        #get rectangle corner points
        p1,p2 = quit.getP1(), quit.getP2()
        # Check if the click is within the rectangle's bounds
        if (p1.getX() <= x <= p2.getX() or p2.getX() <= x <= p1.getX()) and \
           (p1.getY() <= y <= p2.getY() or p2.getY() <= y <= p1.getY()):
            break
        else:
            #draw point
            click.draw(win)
            #increment n count and update sentinals
            n += 1
            xtotal += x
            ytotal += y
            xytotal += x*y
            x2total += x**2

    #process results
    
    #get means
    xbar = xtotal / n
    ybar = ytotal / n

    m = (xytotal - (n * xbar * ybar))/ (x2total - (n * (xbar**2)))

    #draw line - find start and end points

    x0, xfin = -10, 10

    y0 = ybar + m*(x0 - xbar) 
    yfin = ybar + m*(xfin - xbar)       

    regLine = Line(Point(x0, y0), Point(xfin, yfin))
    regLine.draw(win)
    
    win.getMouse()
    win.close()      

main()

 
```

```python
#grayscale

from graphics import *

def main():
    # Prompt for image file
    filename = input("Enter image file name (e.g., image.gif): ")

    # Create window
    win = GraphWin("Grayscale", 800, 800)

    try:
        # Load and display image
        img = Image(Point(400, 400), filename)  # Center image at (400, 400)
        img.draw(win)

        # Wait for mouse click to convert to grayscale
        win.getMouse()

        # Get image dimensions
        width = img.getWidth()
        height = img.getHeight()

        # Loop through each pixel
        for x in range(width):
            for y in range(height):
                # Get RGB values of the pixel
                r, g, b = img.getPixel(x, y)
                # Calculate grayscale brightness using luminance formula
                brightness = int(round(0.299 * r + 0.587 * g + 0.114 * b))
                # Set pixel to grayscale (same value for R, G, B)
                img.setPixel(x, y, color_rgb(brightness, brightness, brightness))

        # Redraw the image to update the display
        img.undraw()
        img.draw(win)

        # Wait for another click to close
        win.getMouse()
        win.close()

    except GraphicsError:
        print("Error: Could not load image. Ensure the file exists and is a valid format (e.g., GIF or PPM).")
        win.close()

main()
```

```python
# event_loop3.py
#      Color changing window with clicks to enter text

from graphics import GraphWin, Entry, Text


def handleKey(k, win):
    if k == "r":
        win.setBackground("pink")
    elif k == "w":
        win.setBackground("white")
    elif k == "g":
        win.setBackground("lightgray")
    elif k == "b":
        win.setBackground("lightblue")


def handleClick(pt, win):
    # create an Entry for user to type in
    entry = Entry(pt, 10)
    entry.draw(win)

    # Go modal: wait until user types <Enter> Key
    while True:
        key = win.getKey()
        if key == "Return":

        # undraw the entry and draw Text
            entry.undraw()
            Text(pt, entry.getText()).draw(win)
            break
            
        elif key == "Escape":
            entry.undraw()
            break
            

    # clear (ignore) any mouse click that occurred during text entry
    win.checkMouse()


def main():
    win = GraphWin("Click and Type", 500, 500)

    # Event Loop: handle key presses and mouse clicks until the user
    #    presses the "q" key.
    while True:
        key = win.checkKey()
        if key == "q":  # loop exit
            break

        if key:
            handleKey(key, win)

        pt = win.checkMouse()
        if pt:
            handleClick(pt, win)

    win.close()


main()
```

```python
from graphics import *
import random
import time

def main():
    win = GraphWin("bounce", 1000, 1000)
    win.setCoords(0, 0, 1000, 1000)
    
    radius = 50
    shape = Circle(Point(500,500), radius)
    shape.setOutline("red")
    shape.setFill("red")
    shape.draw(win)
    
    # Initial random direction (-2 to 2 pixels per frame)
    dx = random.uniform(-2, 2)
    dy = random.uniform(-2, 2)
    
    while True:
        key = win.checkKey()
        if key == "q":  # loop exit
            break
            
        elif key == "c":
            prompt = input("enter new color: ")
            shape.setFill(prompt)
        
        c = shape.getCenter()
        x = c.getX()
        y = c.getY()
        
        # Bounce with some randomness
        if x + radius >= 1000:  # Right edge
            dx = -abs(dx) * random.uniform(0.8, 1.2)  # Reverse and vary
            dy = dy * random.uniform(0.8, 1.2)        # Slightly vary vertical
            
        if x - radius <= 0:     # Left edge
            dx = abs(dx) * random.uniform(0.8, 1.2)
            dy = dy * random.uniform(0.8, 1.2)
            
        if y + radius >= 1000:  # Top edge
            dy = -abs(dy) * random.uniform(0.8, 1.2)
            dx = dx * random.uniform(0.8, 1.2)
            
        if y - radius <= 0:     # Bottom edge
            dy = abs(dy) * random.uniform(0.8, 1.2)
            dx = dx * random.uniform(0.8, 1.2)
        
        shape.move(dx, dy)
        win.update()
        time.sleep(0.001)
        
    win.close()
    
main()
```
