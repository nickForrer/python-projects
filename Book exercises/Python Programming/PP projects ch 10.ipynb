```python
#journey with data from file

def getNumbersFromFile(fname):
    # Robust version that allows multiple numbers per line
    #   and tolerates blank lines
    newnums = []
    with open(fname, "r") as infile:
        for line in infile:
            #print(line)
            tokens = line.split()
            #print(tokens)
            for token in tokens:
                try:
                    num = float(token)  # Try to convert token to float
                    newnums.append(num)
                    #print(num)
                except ValueError:
                    continue  # Skip non-numeric tokens
    return newnums

def main():
    
    totalMiles = 0
    totalGas = 0
    
    data = getNumbersFromFile('gas.txt')
    miles = data[0]
    gastank = data[1]
    
    # Process pairs of numbers (gas, odo) starting from index 2
    for i in range(2, len(data),2):

        gas = data[i]
        odo = data[i+1]
        print("gas: ", gas, "odo: ", odo)
        
        # Calculate gas used and miles traveled
        gasUsed = gastank - gas
        milestrav = odo - miles
        print(f"Gas used: {gasUsed}, Miles traveled: {milestrav}")

        #calc mpg for this leg
        mpg =  milestrav/ gasUsed    
        print(f"The MPG for this leg was {mpg:.2f}")
        
        #update totals
        totalMiles += miles
        totalGas += gasUsed
        
        #Update miles and gastank for the next leg
        miles = odo
        gastank = gas


    print(f"the total MPG was {totalMiles/totalGas:.2f}")
        
main()
```

```python
# Pig Latin translator

def getTextFromFile(fname):
    # Robust version that allows multiple numbers per line
    #   and tolerates blank lines
    text = ''
    with open(fname, "r") as infile:
        for line in infile:
            #print(line)
            words = line.split()
            #print(tokens)
            for word in words:
                    # Check if the word is not an integer
                    if not word.isdigit():
                        # Remove any punctuation if desired (optional)
                        word = ''.join(char for char in word if char.isalpha() or char.isspace())
                        if word:  # Only add non-empty strings
                            text += word + " "
                            
    return text


def firstVowel(word):
    """
    Returns the index of the first vowel (a, e, i, o, u) and the prefix before it.
    """
    vowels = 'aeiouAEIOU'

    for i, char in enumerate(word):
        if char in vowels:
            return i, word[:i]  # Index of vowel, prefix before it
    return len(word), word  # Handle words with no vowels (e.g., "cry")


def transWordPl(word):
    index, rest = firstVowel(word) #Tuple unpacking
    if index == 0:  # Starts with vowel
        return word + "way"
    else:  # Consonant start
        return word[index:] + rest + "ay"  # After vowel + prefix + "ay"

def main():
    prompt = getTextFromFile('pigLatin.txt')
    print(prompt)
    atom = prompt.split()    
    final = ""
    for word in atom:
        final += transWordPl(word) + " "

    print(final)

main()
```

```python
#copies of censored text

#censor directory

from pathlib import Path

def getTextFromFile(fname):
    # Robust version that allows multiple numbers per line
    #   and tolerates blank lines
    text = ''
    with open(fname, "r") as infile:
        for line in infile:
            #print(line)
            words = line.split()
            #print(tokens)
            for word in words:
                    # Check if the word is not an integer
                    if not word.isdigit():
                        # Remove any punctuation if desired (optional)
                        word = ''.join(char for char in word if char.isalpha() or char.isspace())
                        if word:  # Only add non-empty strings
                            text += word + " "
                            
    return text


def getCleanTextFromFile(fname):
    # Robust version that allows multiple numbers per line
    #   and tolerates blank lines
    text = ''
    banned = getTextFromFile('banned.txt').split()
    was_censored = False  # Track if any censoring occurred
    with open(fname, "r") as infile:
        for line in infile:
            words = line.split()
            for word in words:
                # Store original word for case preservation
                modified_word = word
                # Check if any banned word is in the word (case-insensitive)
                for banned_word in banned:
                    if banned_word in word.lower():
                        asterisks = '*' * len(banned_word)
                        # Replace banned word with asterisks, preserving case of rest
                        modified_word = asterisks + word[len(banned_word):]
                        was_censored = True  # Mark that censoring happened
                        break  # Stop checking other banned words
                text += modified_word + " "
    return text, was_censored  # Return text and censoring flag


def getTextFiles():
    files = list(Path(".").glob("*.txt"))
    return files

def main():   
    files = getTextFiles()
    for file in files:
        
        if file.name == 'banned.txt':
            continue
        old = getTextFromFile(file)
        new, was_censored = getCleanTextFromFile(file)
        print(f"Original ({file}): {old}")
        print(f"Censored: {new}\n")
        
        # Only write file if censoring occurred
        if was_censored:
            name = 'censored_' + str(file)
            with open(name, "w") as outfile:
                outfile.write(new)
        else:
            print(f"Skipped {file}: No banned words found.")

main()




```
