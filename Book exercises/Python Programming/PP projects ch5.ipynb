```python
def sumN(n):
    
    total = 0
    for i in range(1,1+n):
        total = total + i
    return total

def sumNCubes(n):
    
    total = 0
    for i in range(1,1+n):
        total = total + i**3
    return total

#sumNCubes(3)


def main():
    var = int(input("please give a num: "))
    result = sumN(var)
    result2 = sumNCubes(var)
    print(f"the sum of integers up to {var} is {result} and the sum of cubes is {result2}")
              
              
main()
```

```python
import math
from graphics import *


def square(x):
    return x * x


def distance(p1, p2):
    dist = math.sqrt(square(p2.getX() - p1.getX())
                     + square(p2.getY() - p1.getY()))
    return dist

def perim(a,b,c):
    return a + b + c

def area(perim, a, b, c):
    s = perim/2
    return math.sqrt(s*((s-a)*(s-b)*(s-c)))

def main():
    win = GraphWin("Draw a Triangle")
    win.setCoords(0.0, 0.0, 10.0, 10.0)
    message = Text(Point(5, 0.5), "Click on three points")
    message.draw(win)

    # Get and draw three vertices of triangle
    p1 = win.getMouse()
    p1.draw(win)
    p2 = win.getMouse()
    p2.draw(win)
    p3 = win.getMouse()
    p3.draw(win)

    # Use Polygon object to draw the triangle
    triangle = Polygon(p1, p2, p3)
    triangle.setFill("peachpuff")
    triangle.setOutline("cyan")
    triangle.draw(win)

    # Calculate the perimeter and area of the triangle
    a, b, c = distance(p1, p2), distance(p2, p3), distance(p3, p1)
    perim1 = perim(a,b,c)
    area1 = area(perim1, a, b, c)
    
    message.setText(f"The perimeter is: {perim1}".format(perim))
    win.getMouse()
                    
    message.setText(f" The area is: {area1}".format(area1))

    # Wait for another click to exit
    win.getMouse()
    win.close()

main()
```

```python
from graphics import *

def drawFace(center, size, win, happy=True):
    # Draw the face (circle)
    face = Circle(center, size)
    face.setFill("yellow")
    face.draw(win)
    
    # Calculate eye positions based on center and size
    eyeSize = size * 0.15
    
    # Left eye
    leftEyeX = center.getX() - size * 0.3
    leftEyeY = center.getY() + size * 0.3
    leftEyeCenter = Point(leftEyeX, leftEyeY)
    leftEye = Circle(leftEyeCenter, eyeSize)
    leftEye.setFill("white")
    leftEye.draw(win)
    
    # Right eye
    rightEyeX = center.getX() + size * 0.3
    rightEyeY = center.getY() + size * 0.3
    rightEyeCenter = Point(rightEyeX, rightEyeY)
    rightEye = Circle(rightEyeCenter, eyeSize)
    rightEye.setFill("white")
    rightEye.draw(win)
    
    # Pupils
    pupilSize = eyeSize * 0.5
    leftPupil = Circle(leftEyeCenter, pupilSize)
    leftPupil.setFill("black")
    leftPupil.draw(win)
    
    rightPupil = Circle(rightEyeCenter, pupilSize)
    rightPupil.setFill("black")
    rightPupil.draw(win)
    
    # Mouth using Arc
    # Define the bounding rectangle for the arc
    mouthWidth = size * 0.6
    mouthHeight = size * 0.3
    mouthLeftX = center.getX() - mouthWidth/2
    mouthRightX = center.getX() + mouthWidth/2
    mouthY = center.getY() - size * 0.3
    
    if happy:
        # Smiley mouth (arc from 0 to 180 degrees)
        mouth = Arc(Point(mouthLeftX, mouthY), Point(mouthRightX, mouthY - mouthHeight), 
                   0, 180)
    else:
        # Unhappy mouth (arc from 180 to 360 degrees)
        mouth = Arc(Point(mouthLeftX, mouthY - mouthHeight), Point(mouthRightX, mouthY), 
                   180, 360)
    
    mouth.setWidth(2)  # Make the line a bit thicker
    mouth.draw(win)

def main():
    win = GraphWin("faces", 500, 500)
    win.setCoords(-10, -10, 10, 10)
    
    p1 = float(input("enter x coordinate of centre: "))
    p2 = float(input("enter y coordinate of centre: "))
    center = Point(p1, p2)
    size = float(input("enter the radius of the face: "))
    mood = input("Happy face? (yes/no): ").lower()
    happy = mood.startswith('y')
    
    drawFace(center, size, win, happy)
    
    win.getMouse()
    win.close()
    
main()
```
