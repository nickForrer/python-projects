```python
# stats.py

from math import sqrt


def getNumbers():
    nums = []     # start with an empty list

    # sentinel loop to get numbers
    xStr = input("Enter a number (<Enter> to quit) >> ")
    while xStr != "":
        x = float(xStr)
        nums.append(x)   # add this value to the list
        xStr = input("Enter a number (<Enter> to quit) >> ")
    return nums


def mean(nums):
    total = 0.0
    for num in nums:
        total = total + num
    return total / len(nums)


def stdDev(nums, xbar):
    sumDevSq = 0.0
    for num in nums:
        dev = num - xbar
        sumDevSq = sumDevSq + dev * dev
    return sqrt(sumDevSq/(len(nums)-1))


def median(nums):
    nums.sort()
    size = len(nums)
    midPos = size // 2
    if size % 2 == 0:
        med = (nums[midPos] + nums[midPos-1]) / 2.0
    else:
        med = nums[midPos]
    return med


def outliers(nums, xbar, s):
    outs = []
    for x in nums:
        if abs((x-xbar)/s) >= 3:
            outs.append(x)
    return outs


def main():
    print("This program computes mean, median and standard deviation.")

    data = getNumbers()
    xbar = mean(data)
    std = stdDev(data, xbar)
    med = median(data)

    print("\nThe mean is", xbar)
    print("The standard deviation is", std)
    print("The median is", med)


if __name__ == '__main__':
    main()
```

```python
#inner product

var1 = [1,2,3]
var2 = [4,5,6]

def matmul(arr1,arr2):
        result = []
        for i,j in zip(arr1, arr2):
             result.append(i*j)
        return sum(result)
    
matmul(var1, var2)
            
```

```python
#inner product without zip or sum

var1 = [1,2,3]
var2 = [4,5,6]

def matmul(arr1,arr2):
        result = []
        for i in range(len(arr1)):
             result.append(arr1[i]*arr2[i])
        total = 0
        for i in result:
            total += i
        
        return total
    
matmul(var1, var2)
```

```python
#Remove duplicate from list

var1 = ['a','b','c', 'b', 'd', 'd','a']

def removeDup(myList):
    seen = []
    for i in myList:
        if i not in seen:
            seen.append(i)
    return seen
                   
removeDup(var1)
```

```python
#prime sieve

def sieve(n):
    nums = [n for n in range(2,n+1)]
    result = []
    for i in nums:
        for j in nums[i:]:
            if j%i == 0:
                #print(j)
                nums.remove(j)
        result.append(i)
        print(f'{i} is prime')
    print(result)
    
sieve(20)
```

```python
#blood pressure

readings = [(120,90),(60,50),(160,100)]

def blood(readings):
    elevated = []
    normal = []
    for i, j in readings:
        if i >= 120 and j >= 80:
            elevated.append(i)
            elevated.append(j)
        else:
            normal.append(i)
            normal.append(j)
    return elevated, normal

blood(readings)
    
```

```python
#histogram

import matplotlib.pyplot as plt
    
def main():
        #data = [1,1,0,3,3,4,9,10,9,4,4,5,6]
        prompt = input("enter a set of scores followed by spaces")
        data = int(prompt.split(" "))
        
        #create list of tuples with number 0-10 as first element and count set to 0 as second
        scores = [(i,0) for i in range(0,11)]
        
        #calculate frequency of the scores
        for num in data:
            for i, (val, count) in enumerate(scores):
                if val == num:
                    scores[i] = (val, count + 1)
                    

        print(scores)

        # Extract numbers and counts
        numbers = [num for num, _ in scores]
        counts = [count for _, count in scores]

        # Plot histogram
        plt.bar(numbers, counts)
        plt.xlabel('Number')
        plt.ylabel('Frequency')
        plt.title('Histogram of Number Frequencies')
        plt.xticks(range(11))
        plt.yticks(range(1, 16))
        plt.show()
        
main()
        
```
