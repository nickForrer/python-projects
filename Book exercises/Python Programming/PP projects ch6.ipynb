```python
#conditional logic


def main():
    
    hours = float(input("how many hrs worked: "))
    rate = float(input("hourly rate: "))
    if hours <= 40:
        pay = hours * rate
        print(f"you earned ${pay:.2f} ")
    else:
        pay = 40 * rate
        extra = hours - 40
        extrapay = extra * (rate* 1.5)
        total = pay + extrapay
        print((f"you earned ${total:.2f} "))

main()
```

```python
def bmi(weight, height):
    result = (weight *703) / (height**2)
    if result in range(19,26):
        print("healthy")
    elif result < 19:
        print("too low")
    elif result > 25:
        print("too high")
        
def main():
    weight = float(input("enter weight in pounds"))
    height = float(input("enter height in inches"))
    bmi(weight, height)
    
main()
```

```python
def speeding(limit, clocked):
    base = 50
    if clocked > limit and clocked < 91:
        fine = base + 5*(clocked-limit)
        print("your fine is", fine)
    elif clocked <= limit:
        print("you were safe")
    elif clocked > 90:
        fine = base + 5*(clocked-limit) + 200
        print("your fine is", fine)
        
def main():
    limit = float(input("speed limit: "))
    clocked = float(input("clocked speed: "))
    speeding(limit, clocked)
    
main()

    
```

```python
def leap(year):
    
    
    
    if year%4 == 0:
        if year%100 != 0: #and (not year%400 ==0):
            return 'leap'
        else:
            if year%400 == 0:
                return 'leap'
            else:
                return 'not leap'
    else:
        return 'not leap'
    
def main():
    year = int(input("enter year: "))
    result = leap(year)
    print(result)
main()
            
```

```python
def valid(date,month,year):

    if date > 31:
        return "not valid"

    elif month in [4,6,9,11] and date > 30:
        return "not valid"

    elif month == 2:
        if year%4 == 0 and (year % 100 != 0 or year % 400 == 0): #century rule
            if date >29:
                return "not valid"
            else:
                return "valid"
        elif date>28:
            return "not valid"
        else:
            return "valid"

    else:
        return "valid"
        
def main():
    
    try:
        result = input("enter date in dd/mm/yyyy format: ")
        var = result.split('/')
        date, month, year = map(int, var)
        result2 = valid(date, month, year)
        print("this date is", result2)
        if result2 == "valid":
            daynum = (31*(month - 1)) + date
            if month>2:
                daynum = daynum - ((4*(month)) + 23)//10
                if year%4 == 0:
                    daynum = daynum +1
            print("the day number is", daynum)

    except (ValueError, IndexError):
        print("not a valid input")

main()




```

```python
#line segment

from graphics import *
import math

def main():

    
    win = GraphWin()
    win.setCoords(-10, -10, 10, 10)
    
    click1 = win.getMouse()
    click2 = win.getMouse()
    
    shape = Line(click1, click2)
    shape.draw(win)
    
    x1 = click1.getX()
    x2 = click2.getX()
    y1 = click1.getY()
    y2 = click2.getY()
    
    dx = x2-x1
    dy = y2-y1
    
    if dx != 0: 
        slope = dy/dx
    
    else:
        slope = "infinite"
    
    length = math.sqrt((dx**2+dy**2))

    mid = Point((x1+x2)/2, (y1+y2)/2)
    mid.setFill("cyan")
    mid.draw(win)
    
    print(slope, length)
    
    win.getMouse()
    win.close()
    
main()
```

```python
#target

def main():
    
    win = GraphWin("bullseye", 800, 800)
    
    
    wshape = Circle(Point(400,400), 150)
    wshape.setOutline("white")
    wshape.setFill("white")
    wshape.draw(win)
    
    bshape = Circle(Point(400,400), 120)
    bshape.setOutline("black")
    bshape.setFill("black")
    bshape.draw(win)
    
    blshape = Circle(Point(400,400), 90)
    blshape.setOutline("blue")
    blshape.setFill("blue")
    blshape.draw(win)
    
    rshape = Circle(Point(400,400), 60)
    rshape.setOutline("red")
    rshape.setFill("red")
    rshape.draw(win)
    
    
    yshape = Circle(Point(400,400), 30)
    yshape.setOutline("yellow")
    yshape.setFill("yellow")
    yshape.draw(win)   
    
   
    
    message = Text(Point(75, 5), "click to close")
    message.draw(win)
    
    total = 0
    for i in range(5):
        shot = win.getMouse()
        # Get shot coordinates
        x = shot.getX()
        y = shot.getY()
        # Distance from center (400, 400)
        distance = math.sqrt((x - 400)**2 + (y - 400)**2)
        
        # Check smallest to largest radius
        if distance <= 30:  # Yellow
            print("9")
            total += 9
        elif distance <= 60:  # Red
            print("7")
            total += 7
        elif distance <= 90:  # Blue
            print("5")
            total += 5
        elif distance <= 120:  # Black
            print("3")
            total += 3
        elif distance <= 150:  # White
            print("1")
            total += 1
        else:
            print("0 - outside target")
            total += 0
    
    print("total score: ", total)
    
    win.getMouse()
    win.close()
    
main()
```

```python
from graphics import *
import random
import time

def main():
    win = GraphWin("bounce", 1000, 1000)
    win.setCoords(0, 0, 1000, 1000)
    
    radius = 50
    shape = Circle(Point(500,500), radius)
    shape.setOutline("red")
    shape.setFill("red")
    shape.draw(win)
    
    # Initial random direction (-2 to 2 pixels per frame)
    dx = random.uniform(-2, 2)
    dy = random.uniform(-2, 2)
    
    for i in range(10000):
        c = shape.getCenter()
        x = c.getX()
        y = c.getY()
        
        # Bounce with some randomness
        if x + radius >= 1000:  # Right edge
            dx = -abs(dx) * random.uniform(0.8, 1.2)  # Reverse and vary
            dy = dy * random.uniform(0.8, 1.2)        # Slightly vary vertical
            
        if x - radius <= 0:     # Left edge
            dx = abs(dx) * random.uniform(0.8, 1.2)
            dy = dy * random.uniform(0.8, 1.2)
            
        if y + radius >= 1000:  # Top edge
            dy = -abs(dy) * random.uniform(0.8, 1.2)
            dx = dx * random.uniform(0.8, 1.2)
            
        if y - radius <= 0:     # Bottom edge
            dy = abs(dy) * random.uniform(0.8, 1.2)
            dx = dx * random.uniform(0.8, 1.2)
        
        shape.move(dx, dy)
        win.update()
        time.sleep(0.001)
        
    win.close()
    
main()
```
