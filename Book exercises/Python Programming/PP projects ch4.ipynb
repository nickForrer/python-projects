```python
#dice
from graphics import *

def main():
    win = GraphWin("Dice", 500, 500)
    win.setCoords(-12, -12, 12, 12)

    # Die 1: Value 1 (one dot in center)
    die1 = Rectangle(Point(-9, 0), Point(-6, 3))
    die1.setFill("white")
    die1.setOutline("black")
    die1.draw(win)
    dot1_1 = Point(-7.5, 1.5)  # Center of die1
    dot1_1.draw(win)

    # Die 2: Value 2 (two dots diagonally)
    die2 = Rectangle(Point(-5, 0), Point(-2, 3))
    die2.setFill("white")
    die2.setOutline("black")
    die2.draw(win)
    dot2_1 = Point(-4.5, 0.5)  # Bottom-left
    dot2_2 = Point(-2.5, 2.5)  # Top-right
    dot2_1.draw(win)
    dot2_2.draw(win)

    # Die 3: Value 3 (three dots in diagonal line)
    die3 = Rectangle(Point(-1, 0), Point(2, 3))
    die3.setFill("white")
    die3.setOutline("black")
    die3.draw(win)
    dot3_1 = Point(-0.5, 0.5)  # Bottom-left
    dot3_2 = Point(0.5, 1.5)   # Center
    dot3_3 = Point(1.5, 2.5)   # Top-right
    dot3_1.draw(win)
    dot3_2.draw(win)
    dot3_3.draw(win)

    # Die 4: Value 4 (four dots in square pattern)
    die4 = Rectangle(Point(3, 0), Point(6, 3))
    die4.setFill("white")
    die4.setOutline("black")
    die4.draw(win)
    dot4_1 = Point(3.5, 0.5)  # Bottom-left
    dot4_2 = Point(5.5, 0.5)  # Bottom-right
    dot4_3 = Point(3.5, 2.5)  # Top-left
    dot4_4 = Point(5.5, 2.5)  # Top-right
    dot4_1.draw(win)
    dot4_2.draw(win)
    dot4_3.draw(win)
    dot4_4.draw(win)

    # Die 5: Value 5 (four corners plus center)
    die5 = Rectangle(Point(7, 0), Point(10, 3))
    die5.setFill("white")
    die5.setOutline("black")
    die5.draw(win)
    dot5_1 = Point(7.5, 0.5)  # Bottom-left
    dot5_2 = Point(9.5, 0.5)  # Bottom-right
    dot5_3 = Point(7.5, 2.5)  # Top-left
    dot5_4 = Point(9.5, 2.5)  # Top-right
    dot5_5 = Point(8.5, 1.5)  # Center
    dot5_1.draw(win)
    dot5_2.draw(win)
    dot5_3.draw(win)
    dot5_4.draw(win)
    dot5_5.draw(win)

    win.getMouse()
    win.close()

main()
```

```python
#futval

from graphics import *


def main():
    # Introduction
    print("This program plots the growth of a 10-year investment.")

    

    # Create a graphics window with labels on left edge
    win = GraphWin("Investment Growth Chart", 850,850)
    win.setBackground("white")
    win.setCoords(-1.75, -200, 11.5, 10400)
    Text(Point(-1, 0), ' 0.0K').draw(win)
    Text(Point(-1, 2500), ' 2.5K').draw(win)
    Text(Point(-1, 5000), ' 5.0K').draw(win)
    Text(Point(-1, 7500), ' 7.5k').draw(win)
    Text(Point(-1, 10000), '10.0K').draw(win)
    
    # Get principal and interest rate   
    
    # Create entry boxes for principal and APR
    #Text(Point(3, 10000), "Initial Principal:").draw(win)
    principal1 = Entry(Point(6, 10000), 50)
    principal1.setText("Initial Principal: ")  
    principal1.draw(win)

    #Text(Point(3, 9500), "APR (e.g., 0.05):").draw(win)
    apr1 = Entry(Point(6, 9500), 50)
    apr1.setText("APR: ")  # Default value
    apr1.draw(win)
    
    win.getMouse()
    
    # Extract numbers after the prompts
    principal_text = principal1.getText().split()  # e.g., "Initial Principal: 1000"
    apr_text = apr1.getText().split()              # e.g., "APR: 0.05"

    # Split and get the number part
    principal = float(principal_text[-1])
    apr = float(apr_text[-1])

    
    # Draw bar for initial principal
    bar = Rectangle(Point(0, 0), Point(1, principal))
    bar.setFill("green")
    bar.setWidth(2)
    bar.draw(win)

    # Draw a bar for each subsequent year
    for year in range(1, 11):
        principal = principal * (1 + apr)
        bar = Rectangle(Point(year, 0), Point(year+1, principal))
        bar.setFill("green")
        bar.setWidth(2)
        bar.draw(win)

    input("Press <Enter> to quit.")
    win.close()

main()
```

```python
#circle intersect

from graphics import *
import math

def main():
    
    r = float(input("radius: "))
    y = float(input("y intercept: "))
    
    win = GraphWin()
    win.setCoords(-10, -10, 10, 10)
    
    shape = Circle(Point(0,0), r)
    shape.setOutline("black")
    shape.setFill("white")
    shape.draw(win)
    
    
    shape2 = Line(Point(-10,y), Point(10,y))
    shape2.draw(win)
    
    
    x1 = math.sqrt((r**2)-(y**2))
    x2 = -math.sqrt((r**2)-(y**2))
    print(x1, x2)
    
    var1 = Point(x1, y)
    var2 = Point(x2, y)
    
    var1.setFill("red")
    var2.setFill("red")
    
    var1.draw(win)
    var2.draw(win)
    
    
    win.getMouse()
    win.close()
    
main()
    
    
```

```python
#line segment

from graphics import *
import math

def main():

    
    win = GraphWin()
    win.setCoords(-10, -10, 10, 10)
    
    click1 = win.getMouse()
    click2 = win.getMouse()
    
    shape = Line(click1, click2)
    shape.draw(win)
    
    x1 = click1.getX()
    x2 = click2.getX()
    y1 = click1.getY()
    y2 = click2.getY()
    
    dx = x2-x1
    dy = y2-y1
    
    if dx != 0: 
        slope = dy/dx
    else:
        slope = "infinite"
    
    length = math.sqrt((dx**2+dy**2))

    mid = Point((x1+x2)/2, (y1+y2)/2)
    mid.setFill("cyan")
    mid.draw(win)
    
    print(slope, length)
    
    win.getMouse()
    win.close()
    
main()
```

```python
#rectangle

from graphics import *
import math

def main():

    
    win = GraphWin()
    win.setCoords(-10, -10, 10, 10)
    
    click1 = win.getMouse()
    click2 = win.getMouse()
    
    shape = Rectangle(click1, click2)
    shape.draw(win)
    
    x1 = click1.getX()
    x2 = click2.getX()
    y1 = click1.getY()
    y2 = click2.getY()
    
    length = y2 - y1
    width = x2 - x1
    area = length*width
    perimeter =2*(length + width)
    
    print(area, perimeter)
    
    win.getMouse()
    win.close()
    
main()
```

```python
#Triangle

from graphics import *
import math

def main():

    
    win = GraphWin("Triangle", 500,500)
    win.setCoords(-10, -10, 10, 10)
    
    click1 = win.getMouse()
    click2 = win.getMouse()
    click3 = win.getMouse()
    
    shape = Polygon(click1, click2, click3)
    shape.draw(win)
    
    x1 = click1.getX()
    x2 = click2.getX()
    x3 = click3.getX()
    y1 = click1.getY()
    y2 = click2.getY()
    y3 = click3.getY()
    
    dx1 = x2-x1
    dy1 = y2-y1
    dx2 = x3-x2
    dy2 = y3-y2
    dx3 = x3-x1
    dy3 = y3-y1
  
    
    a = math.sqrt((dx1**2+dy1**2))
    b = math.sqrt((dx2**2+dy2**2))
    c = math.sqrt((dx3**2+dy3**2))
    
    perimeter = a + b + c
    s = perimeter/2
    area = math.sqrt(s*((s-a)*(s-b)*(s-c)))
    
    
    print(area, perimeter)
    
    win.getMouse()
    win.close()
    
main()
```
