```python
#to open a file use open(filename.ext[,mode])

#ex

f = open('hello.py')
filecontents = f.read()
print(filecontents)
f.close()
```

```python
#using a context manager keyword 'with'

with open('hello.py') as f:
    filecontents = f.read()
    print(filecontents)
    
#The unindented line below is outside the with block
print('File is closed ' ,'\n', f.closed)
```

```python
#other kinds of file e.g. jpeg need to rb i.e. read binary

with open('happy_pickle.jpeg', 'rb') as f:
    filecontents = f.read()
    print(filecontents)

```

```python
#Reading files

#3 choices of method read([size]), readline() and readlines() - last two only good for text files

with open('quotes2.txt') as f:
    #read in entire file
    content = f.read()
    print(content)

```

```python
with open('quotes2.txt') as f:
    #reads all lines first, then loops through
    for one_line in f.readlines():
        print(one_line, end ='')
```

```python
with open('quotes2.txt') as f:
    #reads all lines first, then loops through
    #count each line starting at 0
    for one_line in enumerate(f.readlines()):
        #if counter is even number, print with no extra newline
        if one_line[0] % 2 == 0:
            print(one_line[1], end ='')
            
        #otherwise print a couple of spaces and a new line
        else:
            print('  ' + one_line[1])
```

```python
#read one line at a tiem if youre unsure of the size of the file

with open('quotes2.txt') as f:
    #reads all lines first, then loops through
    one_line = f.readline()
    while one_line:
        print(one_line, end = '')
        one_line = f.readline()
```

```python
#to format the above

#store a number to use as a loop counter

counter = 1

#open the file

with open('quotes2.txt') as f:
    #read one line from the file
    one_line = f.readline()
    #as long as there are lines to read
    while one_line:
        #if teh counter is an even number, print a couple of spaces
        if counter % 2 == 0:
            print('  ' + one_line)
        #otherwise print with no new line at the end
        else:
            print(one_line, end = '')
        #increment the counter
        counter += 1
        #read teh next line
        one_line = f.readline()
```

```python
#modifying files

#write vs append

new_name = 'Pena Calderon\n'

with open('names.txt', 'a') as f:
    f.write(new_name)
    
print('\nDone')

#reopen the file

with open('names.txt') as f:
    print(f.read())
```

```python
#using tell() to fnd pointer

with open('names.txt') as f:
    print(f.tell())
    one_line = f.readline()
    while one_line:
        
        print(one_line[:-1], f.tell())
        one_line = f.readline()
```

```python
#READING AND COPYING A BINARY FILES - note these may be huge so break file up using following code

#specify the file to copy
file_to_copy = 'happy_pickle.jpeg'
#create new file name with copy before the xtension
name_parts = file_to_copy.split('.')
new_file = name_parts[0] + '_copy .' + name_parts[1]
#open the orignial file as read only binary
with open(file_to_copy, 'rb') as original_file:
    #create or open file to copy into
    with open(new_file, 'wb') as copy_to:
        #grab a chunk of the original file (4MB)
        chunk = original_file.read(4096)
        #loop through until no more chunks
        while len(chunk) > 0:
            copy_to.write(chunk)
            #make sure you read the next chunk in this loop
            chunk = original_file.read(4096)
            
#close is automatic after loops, show done message
print('Done!')
```
