```python
#functions

def alphabetize(original_list=[]):
    '''Pass in any list in square brackets, displays 
    a string with items sorted '''
    #inside the func make a working copy of the list passed in
    sorted_list = original_list.copy()
    #sort the working copy
    sorted_list.sort()
    #make a new empty string for output
    final_list = ''
    #loop through sorted list and append name and comma and space
    for name in sorted_list:
        final_list += name + ', '
    #knock off last comma space if the string is not blank
    final_list = final_list[:-2]
    #print the alphabetized list
    print(final_list)
    
    
alphabetize(['Johnson', 'Maier', 'Tate', 'Sartorial'])
    
```

```python
names = ['Johnson', 'Maier', 'Tate', 'Sartorial']

alphabetize(names)
```

```python
#anonymous/lamda funcs

#lamda arguments: expression

names = ['Adams', 'Ma', 'diMeola', 'Zandusky']
names.sort()
print(names)
```

```python
#want to ignore lowercase - so make func

def lowercaseof(anystring):
    ''' converts string to all lower case'''
    return anystring.lower()

lowercaseof('Zandusky')
```

```python
names.sort(key = lowercaseof)
print(names)
```

```python
lambda anystring: anystring.lower()

names.sort(key = lambda anystring: anystring.lower())

print(names)
```
