```python
#creating/raising your own exceptions



#base class for defining your own user defined exceptions
class Error(Exception):
    '''Base class for other exceptions'''
    pass


#now define your exception as a subclass or error
class EmptyFileError(Error):
    pass




print('do this first')

try:
    
    viewfile = open('people.csv')
    print(viewfile.name)
    
    line_count = len(viewfile.readlines())
    
    if line_count < 2:
        
        raise EmptyFileError
    
    
    #viewfile.wookems()
    
#handles missing file error
except FileNotFoundError:
    print('sorry I cant find file')


#handles my custom error for too few rows
except EmptyFileError:
    print('your file doesnt have enough stuff')
    
#handles all other exceptions
except Exception as err:
    print('\n\nFailed: the error was ' + str(err))
    
else:
    print('show this if there is no exception!')
    
    for oneline in viewfile:
        
        print(oneline)
    
finally:
    
    print('this is in the finally section')
    
print('this is outside the try except, else, finally')


```
