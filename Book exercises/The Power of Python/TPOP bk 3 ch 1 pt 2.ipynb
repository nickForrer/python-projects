```python
#converting strings

import csv
#Open CSV file with ut8 encoding, dont read in newline char
with open('sample.csv', encoding = 'utf-8', newline='') as f:
    #create a CSV row counter and row reader
    #reader method reads one row at a time
    reader = enumerate(csv.reader(f))
    #Loop through one row at a time, i is counter row is entire row
    for i, row in reader:
        #row 0 is just column headings, ignore it
        if i > 0:
            #whole name split into two at comma
            try:
                full_name = row[0].split(',')
                #last name, strip extra spaces.
                last_name=full_name[0].strip()
                #first name, strip extra spaces.
                first_name=full_name[1].strip()
            except IndexError:
                full_name = last_name = first_name = ""
            print(first_name, last_name)
print('Done!')
```

```python
#Converting to Integers and date and Boolean and Floats

import csv
import datetime as dt
#Open CSV file with ut8 encoding, dont read in newline char
with open('sample.csv', encoding = 'utf-8', newline='') as f:
    #create a CSV row counter and row reader
    #reader method reads one row at a time
    reader = enumerate(csv.reader(f))
    #Loop through one row at a time, i is counter row is entire row
    for i, row in reader:
        #row 0 is just column headings, ignore it
        if i > 0:
            #whole name split into two at comma
            try:
                full_name = row[0].split(',')
                #last name, strip extra spaces.
                last_name=full_name[0].strip()
                #first name, strip extra spaces.
                first_name=full_name[1].strip()
            except IndexError:
                full_name = last_name = first_name = ""
            #Birth year as integer, zero for empty string
            birth_year = int(row[1] or 0)
            #date joined is a date
            try:
                date_joined = dt.datetime.strptime(row[2], "%m/%d/%Y").date()
            except ValueError:
                date_joined = None
            # is_active is a Boolean, automatically false for empty string
            is_active = (bool(row[3]))
            # remove $, commas, leading trail spaces so can do math ops
            str_balance = (row[4].replace('$', '').replace(',', '')).strip()
            #balance is a float or 0 for empty string
            balance = float(str_balance or 0)
            print(first_name, last_name, birth_year, date_joined, is_active, balance)
print('Done!')
```

```python
#converting to object/classes

import datetime as dt
import csv
#use these funcs to convert any string to app python date type

#get just the first name from full name
def fname(any):
    try:
        nm = any.split(',')
        return nm[1]
    except IndexError:
        return ''
#get just the last name
def lname(any):
    try:
        nm = any.split(',')
        return nm[0]
    except IndexError:
        return ''
#convert str to int or 0 if no value
def integer(any):
    return int(any or 0)
#convert to date or none if no valid date
def date(any):
    try:
        
        return dt.datetime.strptime(any,"%m/%d/%Y").date()
    except ValueError:
        return None
    
#convert to boolean
def boolean(any):
    return bool(any)

#convert to float
def floatnum(any):
    s_balance = (any.replace('$', '').replace(',', ''))
    return float(s_balance or 0)

#create a list of people

people = []

#define a class where each person is an object

class Person:
    def __init__(self, id, first_name, last_name, birth_year, date_joined, is_active, balance):
        self.id = id
        self.first_name = first_name
        self.last_name = last_name
        self.birth_year = birth_year
        self.date_joined = date_joined
        self.is_active = is_active
        self.balance = balance
        
#open csv file

with open('sample.csv', encoding = 'utf-8', newline='') as f:
    #create a CSV row counter and row reader
    #reader method reads one row at a time
    reader = enumerate(csv.reader(f))
    #Loop through one row at a time, i is counter row is entire row
    f.readline()
    for i, row in reader:
        people.append(Person(i, fname(row[0]), lname(row[0]), integer(row[1]), date(row[2]),
                             boolean(row[3]), floatnum(row[4])))
        
        
for p in people:
    print(p.id, p.first_name, p.last_name, p.birth_year, p.date_joined, p.is_active, p.balance)
        

```
