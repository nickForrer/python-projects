```python
#dictonaries

#use case access and retrieve data values using a key value pair

#syntax dict = {key:value, key:value etc}

#use index[] or get() method to retieve value

#lso update method

people = {'Homesec': 'Pritti Patel', 'PM': 'Boris Johnson'}
print(people)
```

```python
people.update({'Homesec': 'Suela Braviman'})
print(people)
```

```python
people.update({'chancellor': 'Rishi Sunak'})
print(people)
```

```python
for person in people.keys():
    print(person + ' = ' + people[person])
```

```python
for person in people.values():
    print(person)
```

```python
for key, value in people.items():
    print(key, ' = ', value)
```

```python
del people['Homesec']
print(people)
```

```python
product = {
    'name': 'RayBan',
    'unit price': 112.99,
    'taxable': True,
    'in_stock': 10,
    'models': ['black', 'tortoise']
}

for name, item in product.items():
    print(name + ': ', item)
    
```
