```python
#timedelta object for timespans/duration

import datetime as dt
new_years_day = dt.date(2019, 1, 1)
memorial_day = dt.date(2019, 5, 27)
days_between = memorial_day - new_years_day
print(days_between)
```

```python
now = dt.datetime.now()
birthdatetime = dt.datetime(1979, 6, 13)
age = now - birthdatetime
print(age)
print(type(age))
```

```python
today = dt.date.today()
birthdate = dt.date(1979, 6, 13)
delta_age = (today - birthdate)
print(delta_age)
days_old = delta_age.days
print(days_old)
years_old = days_old // 365
print(years_old)
```
