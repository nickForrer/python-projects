```python
import json

#this is the excel data (no keys)
filename = 'people_from_excel.json'
#open the file
with open(filename, 'r', encoding = 'utf-8', newline='') as f:
    #load the whole json file into an obj named people
    people = json.load(f)
    #print(people)
    #print(type(people))
    for p in people:
        #print(type(p))
        print(p['Full Name'], p['Birth Year'], p['Date Joined'], p['Is Active'], p['Balance'])
```

```python
import json
import datetime as dt

#this is the excel data (no keys)
filename = 'people_from_excel.json'
#open the file
with open(filename, 'r', encoding = 'utf-8', newline='') as f:
    #load the whole json file into an obj named people
    people = json.load(f)
    #print(people)
    #print(type(people))
    for p in people:
        #print(type(p))
        name = p['Full Name']
        byear = p['Birth Year']
        joined = p['Date Joined']
        balance = p['Balance']
        print( f"{name: <22}  {byear} {joined:>10} {balance:>12}")
```

```python
#converting whole json file into one string then iterating through it

import json

json_string = """
{
"people": [
 {
   "Full Name": "Angst, Annie",
   "Birth Year": 1982,
   "Date Joined": "01/11/11",
   "Is Active": true,
   "Balance": "$300.00"
 },
 {
   "Full Name": "Bon, Barry",
   "Birth Year": 1973,
   "Date Joined": "02/11/12",
   "Is Active": false,
   "Balance": "-$123.45"
 },
 {
   "Full Name": "Sch, Sandy",
   "Birth Year": 2004,
   "Date Joined": "03/03/03",
   "Is Active": true,
   "Balance": "$0.00"
 }
]
}
"""
peep_data = json.loads(json_string)
for p in peep_data['people']:
    print(p['Full Name'], p['Birth Year'], p['Date Joined'], p['Is Active'], p['Balance'])
```
